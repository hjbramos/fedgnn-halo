# Comparação de métodos e agrupamentos

Execuções curtas em dados sintéticos. Os resultados ilustram a API e não permitem tirar conclusões sobre mobilidade ou sobre superioridade dos métodos.


```python
from pathlib import Path
import tempfile
import copy
import yaml
import pandas as pd
from fedgnn_halo.cli import run

root = Path.cwd()
if not (root / "examples").exists():
    root = root.parent
base = yaml.safe_load((root / "examples/synthetic.yaml").read_text())
output = Path(tempfile.mkdtemp(prefix="halo-comparison-"))
```

```python
rows = []
for partition in ["contiguous", "random"]:
    for method in ["centralized", "local", "fedavg", "fedprox", "ditto", "halo"]:
        config = copy.deepcopy(base)
        config["method"] = method
        config["partition"]["method"] = partition
        config["output"] = str(output / partition / method)
        result = run(config)
        rows.append({"partition": partition, "method": method, "MAE": result["test"]["MAE"]})
pd.DataFrame(rows)
```

O centralizado não usa o agrupamento no treinamento e funciona como referência. Para um estudo empírico, defina previamente sementes, orçamento, critérios de comparação e acesso aos dados.
