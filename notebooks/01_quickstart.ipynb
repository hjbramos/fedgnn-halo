Este exemplo usa uma série sintética e CPU. Verifica o funcionamento do treinamento, não reproduz os resultados da dissertação. Instale o projeto com `pip install -e ".[notebooks]"` antes de abrir este notebook.


```python
from pathlib import Path
import tempfile
import yaml
from fedgnn_halo.cli import run

root = Path.cwd()
if not (root / "examples").exists():
    root = root.parent
options = yaml.safe_load((root / "examples/synthetic.yaml").read_text())
options
```

## Configuração

Altere `model`, `partition` e `training`. O FedGNN-Halo (`halo`) exige LEConv; para GRU ou LSTM use `fedavg`, `fedprox`, `ditto`, `local` ou `centralized`.


```python
options["output"] = str(Path(tempfile.mkdtemp(prefix="halo-notebook-")) / "run")
result = run(options)
result["test"]
```

```python
import pandas as pd
import matplotlib.pyplot as plt

history_path = next(Path(options["output"]).rglob("history.csv"))
history = pd.read_csv(history_path)
history.plot(x="round", y="validation_MAE")
plt.xlabel("Rodada")
plt.ylabel("MAE de validação (unidade da série sintética)")
history
```

Os históricos, a configuração resolvida, as versões e as métricas ficam no diretório informado em `output`. O resultado de teste corresponde ao modelo selecionado pela validação.
